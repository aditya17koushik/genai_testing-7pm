# Why Do We Need a Loop?

Imagine your manager asks you to send a reminder email to **500 employees**.

Would you write the same code 500 times?

No.

Instead, we write the code **once** and let Python repeat it for every employee.

This is exactly why loops exist.

A loop helps us **automate repetitive tasks**.

# 🎯 There are Two Types of Loops in Python

Python provides two loops.

1️⃣ While Loop

Use a while loop when you don't know exactly how many times something should repeat.

Example:

Keep filling a bottle until it becomes full.

You don't know whether it'll take 5 seconds or 10 seconds.

You just keep going until the condition becomes false.

# 2️⃣ For Loop

Use a for loop when you already know how many times you want to repeat something.

Example:

Take attendance of 40 students.

You already know there are 40 students.

So you repeat the same action exactly 40 times.

# 📖 What is a While Loop?

A while loop repeats a block of code as long as a condition is True.

The moment the condition becomes False, the loop stops.

Think of it as asking a question again and again.

"Should I continue?"

If the answer is Yes, continue.

If the answer is No, stop.

# What is a For Loop?

A **for loop** is used to repeat a block of code for every item in a collection.

Think of it like this:

> "For every item, perform the same action."

Examples:

- For every employee, calculate salary.
- For every student, calculate marks.
- For every file, read the contents.
- For every product, calculate GST.

Instead of writing the same code multiple times, Python repeats it automatically.

# Real-Life Examples

Suppose you have:

- 100 employees
- 500 customers
- 1000 invoices
- 50 files
- 200 API responses

In all these situations, you need to perform **the same operation** on every item.

A `for` loop helps us do this easily.

# Syntax

```python
for variable in iterable:
    statements
```

Every for loop has three parts:

- `for` → Python keyword
- `variable` → Holds one item at a time
- `iterable` → The collection we want to process

# What is an Iterable?

An iterable is simply a collection of items.

Examples include:

- List
- Tuple
- Set
- Dictionary
- String
- Range

Python takes **one item at a time** from the iterable until there are no items left.

# Understanding the Loop Variable

Example:

```python
for employee in employees:
```

Here,

`employee` represents **one employee at a time**.

First iteration

employee = First Employee

Second iteration

employee = Second Employee

Third iteration

employee = Third Employee

The variable name can be anything meaningful.

Examples:

- employee
- student
- customer
- product
- file

# What is range()?

`range()` is a built-in Python function used to generate numbers.

It is commonly used when we want to repeat something a fixed number of times.

Example:

Repeat a task 5 times.

Instead of writing the task 5 times, we use `range(5)`.

````python
for i in range(100, 0, -1):
    print(i)

print("Countdown complete!")
````

# Where Do We Use For Loops?

For loops are used almost everywhere in software development.

Examples:

- Reading data from Excel files
- Processing CSV files
- Reading log files
- Processing employee records
- Validating API responses
- Reading database records
- Processing customer orders
- Generating reports

Whenever the same task needs to be performed on multiple items, a for loop is a good choice.

# When Should We Use a For Loop?

Use a for loop when:

- You have multiple items.
- The same operation needs to be performed on every item.
- You want Python to automatically move from one item to the next.

If you are processing a collection, a `for` loop is usually the right choice.

````python
employees = ["John", "Alice", "David", "Emma", "Michael"]

for i in employees:
    print(i)
````
