# DICTIONARY

```python
#creating a dict
ice_cream={"flavour_1":"Vanilla",
           "flavour_2":"Chocolate",
           "flavour_3":"butterscotch"}
print(type(ice_cream))
```

```python
#read/retrieve
print(ice_cream['flavour_1'])
print(ice_cream['flavour_2'])
print(ice_cream['flavour_3'])
```

```python
print(ice_cream.get("flavour_1"))
print(ice_cream.get("flavour_2"))
print(ice_cream.get("flavour_3"))
```

```python
ice_cream.keys()
```

```python
ice_cream.values()
```

```python
ice_cream.items()
```

```python
#update
ice_cream["flavour_4"] ="oreo"
ice_cream
```

```python
ice_cream.update({"flavour_5":"strawberry"})
ice_cream
```

```python
#remove
ice_cream.pop("flavour_5")
```

#SET---> UNIQUE, and only stores immutable elements

```python
#create
profile={1,2,3,3}
print(profile)
print(type(profile))
```

```python
profile_2={3,4,"hii",(1,2,3)}
print(type(profile_2))
profile_2
```

```python
#union operation
union=profile.union(profile_2)

union
```

```python
union_1= profile | profile_2
union_1
```

```python
#intersection operation
intersection=profile.intersection(profile_2)
intersection
```

```python
intersection_2= profile & profile_2
intersection_2
```

```python
set_1={1,2,3,4,"huiu"}
set_2={2,3,4,5,"hahah"}
print(set_1)
print(set_2)
```

```python
difference=set_1-set_2
difference
```

```python
difference_2=set_2-set_1
difference_2
```

```python
#add element in set
print(set_1)
set_1.add((0,0,0))
set_1
```

```python
#update element
set_1.update("0.1hii")
set_1.update([0.1,0.2])
set_1
```
