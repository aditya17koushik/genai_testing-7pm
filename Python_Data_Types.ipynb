# Integer (`int`)

## Definition

## Why?
Used to store this kind of information efficiently.

## Properties
- Built-in datatype
- Check using `type()`

## Real-world Example
Think of an application storing this value.

## Practice

# String Operations
- Concatenation
- Indexing
- Slicing

````
s1="Hello"
s2="World"
print(s1+" "+s2)
print(s1[0], s1[-1])
print(s1[1:4])
````

# List Operations
Create lists with mixed data types, access, append, update, remove

````
lst=[1,"Python",3.14,True]
print(lst)
print(lst[1])
lst.append("AI")
lst[0]=100
lst.remove(True)
print(lst)
````

# Tuple Operations

````
t=(1,"AI",3.5)
print(t)
print(t[1])
````

# Dictionary Operations

````
d={"name":"Alice","age":20}
print(d["name"])
d["age"]=21
d["city"]="Hyderabad"
d.pop("name")
print(d)
````


# Memory Representation

### String
```text
s = "CAT"
+---+---+---+
| C | A | T |
+---+---+---+
 0   1   2
```

### List
```text
list
+---+---+---+
| * | * | * | ---> objects
+---+---+---+
```

### Tuple
```text
tuple (immutable)
+---+---+---+
| * | * | * |
+---+---+---+
```

### Dictionary
```text
Hash Table
key ------> value
"name" ---> "Alice"
"age" ----> 20
```
